### Catalog at external location

```python
%sql
create catalog if not exists `dev_ext`
managed location 'abfss://unitycontainer@azstorageadls.dfs.core.windows.net/catalog'
comment 'This is an external catalog'
```

### Schema at external location


```python
%sql
create catalog if not exists `dev_schema_ext`
managed location 'abfss://unitycontainer@azstorageadls.dfs.core.windows.net/schema'
comment 'This is an external schema'
```

```python
%sql
describe catalog extended sql_dev;
```

```python
%sql
create schema if not exists dev.bronze
comment 'This schema is under dev managed catalog';
```

```python
%sql
create schema if not exists dev_ext.bronze_ext
comment 'This schema is under dev external catalog'
managed location 'abfss://unitycontainer@azstorageadls.dfs.core.windows.net/catalog/bronze_ext';
    
```

### CREATE TABLES


```python
%sql
CREATE TABLE IF NOT EXISTS dev.bronze.RAW_SALE
(
  INVOICE_id STRING,
  CUST_ID STRING,
  PRODUCT_CODE STRING,
  QUANTITY DOUBLE,
  PRICE DOUBLE
);

INSERT INTO  dev.bronze.RAW_SALE VALUES('INV001','CUST1001','PCODE1001',10,100);
```
