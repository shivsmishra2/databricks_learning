```python
%sql
CREATE OR REFRESH STREAMING TABLE dev.bronze.emp_st
AS
SELECT * FROM 
STREAM read_files(
  "/Volumes/dev/bronze/ext_volume/files/",
  format=>'csv'
)
```

```python
%sql
describe extended dev.bronze.emp_st;
```

```python
%sql
select * from dev.bronze.emp_st;
```

```python
%sql
-- Exclude the processed files:
CREATE OR REFRESH STREAMING TABLE dev.bronze.emp_st
SCHEDULE EVERY 4 HOUR
AS
SELECT * FROM 
STREAM read_files(
  "/Volumes/dev/bronze/ext_volume/files/",
  format=>'csv',
  includeExistingFiles => false,
  ignoreMissingFiles => true 
)
```

```python
%sql
--- Incremental refresh option:
REFRESH STREAMING TABLE dev.bronze.emp_st;

-- FULL Refresh
refresh streaming table dev.bronze.emp_st FULL;
    
select * from dev.bronze.emp_st;
```
