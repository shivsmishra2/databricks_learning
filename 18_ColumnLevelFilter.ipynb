```python
%sql
--masking function dev.metadata.phone_mask
CREATE OR REPLACE FUNCTION dev.metadata.phone_mask(col STRING)
RETURNS STRING
LANGUAGE SQL
RETURN(
select case when group = 'admin' THEN phone_mask.col ELSE '***-***-***' end as masked_val from(
select  group from dev.metadata.user_groups WHERE user_email = current_user() limit 1)
)
```

```python
%sql
-- validate
select  dev.metadata.phone_mask('889282828')
```

```python
%sql
-- mask the column
alter table dev.etl.customer_raw alter c_phone set mask dev.metadata.phone_mask;
```
