```python
%sql
--- CREATE SCALAR FUNCTION TO CALCULATE TAX SQL
CREATE OR REPLACE FUNCTION dev.bronze.tax_sql(sal double)
RETURNS double
LANGUAGE SQL
RETURN sal * 0.3
```

```python
%sql
--- CALLING ABOVE FUNCTION
select dev.bronze.tax_sql(100000)
```

```python
%sql
select *, dev.bronze.tax_sql(UnitPrice) as tax_sql from dev.bronze.sales
```

```python
##  CALLING SQL UDF USING PYTHON
from pyspark.sql.functions import expr
df = spark.read.table("dev.bronze.sales")
df.withColumn("tax",expr("dev.bronze.tax_sql(unitPrice)")).display()
```

```python
%sql
-- SCALAR FUNCTION FOR API

CREATE OR REPLACE FUNCTION dev.bronze.fruit_nutrition (fruit_name STRING) RETURNS STRING
LANGUAGE PYTHON
AS
$$
import requests
api_url = f"https://www.fruityvice.com/api/fruit/{fruit_name}"
response = requests.get(api_url)
data = response.json()
return str(data.get('nutritions', 'NA'))
$$
;
```

```python
%sql
select dev.bronze.fruit_nutrition('apple')
  
```

### TABULAR FUNCTION


```python
%sql
CREATE OR REPLACE FUNCTION dev.bronze.sales(customer_id INT)
RETURNS TABLE (InvoiceNo STRING, StockCode STRING, Description STRING)
LANGUAGE SQL
RETURN(
  select InvoiceNo,StockCode,Description from dev.bronze.sales where CustomerID = customer_id
)

--select * from dev.bronze.sales LIMIT 2
```
